## Description générale sur dataset

```python

RGPH2024 = pd.ExcelFile("Indicateurs démographiques et socioéconomiques du Royaume du Maroc selon les résultats du RGPH 2024.xlsx")

# Function to clean MultiIndex columns (fix Unnamed, NaN)
def clean_multilevel_columns(df):
    new_cols = []
    for col in df.columns:
        clean_col = tuple(
            "" if ("Unnamed" in str(c) or pd.isna(c)) else str(c).strip()
            for c in col
        )
        new_cols.append(clean_col)
    df.columns = pd.MultiIndex.from_tuples(new_cols)
    return df


def forward_fill_column_levels(df):
    df.columns = pd.MultiIndex.from_tuples([
        tuple(pd.Series(col).ffill()) for col in df.columns
    ])
    return df

# Read sheets with 3 header rows
population_totale = pd.read_excel(RGPH2024, sheet_name="Population", header=[0,1,2])
population_urbaine = pd.read_excel(RGPH2024, sheet_name="Population_Urbaine", header=[0,1,2])
population_rurale = pd.read_excel(RGPH2024, sheet_name="Population_Rurale", header=[0,1,2])

# Clean and fix headers
population_totale = clean_multilevel_columns(population_totale)
population_totale = forward_fill_column_levels(population_totale)

population_urbaine = clean_multilevel_columns(population_urbaine)
population_urbaine = forward_fill_column_levels(population_urbaine)

population_rurale = clean_multilevel_columns(population_rurale)
population_rurale = forward_fill_column_levels(population_rurale)

# Check columns for population total
for col in population_totale.columns:
    print(col)

```

```python
# Flatten columns (tuples → strings séparées par un _)
def flatten_columns(df, sep="_"):
    df.columns = [sep.join(filter(None, map(str, col))).strip() for col in df.columns]
    return df

population_totale = flatten_columns(population_totale)
population_urbaine = flatten_columns(population_urbaine)
population_rurale = flatten_columns(population_rurale)
```

```python
# Nettoyage des codes géographiques (suppression des .0)
def clean_code_geo(df):
    df = df.copy()
    df['Code_geographique'] = df['Code_geographique'].astype(str).str.replace('.0', '', regex=False)
    return df

population_all = clean_code_geo(population_all)
menages_all = clean_code_geo(menages_all)
```

```python
def filter_regions_and_total(df):
    return df[
        (df['Collectivites_territoriales'] == 'Ensemble du territoire national') |
        (df['Code_geographique'].str.match(r'^\d{1,2}$', na=False))  # Ex : 1, 11
    ]

population_regions = filter_regions_and_total(population_all)
menages_regions = filter_regions_and_total(menages_all)
```

```python
population_regions
```

```python
population_regions.loc[population_regions['Code_geographique'] == 'nan', 'Code_geographique'] = 0
```

```python
menages_regions.loc[menages_regions['Code_geographique'] == 'nan', 'Code_geographique'] = 0
menages_regions
```

```python
def sort_by_code_geo(df):
    df = df.copy()
    df['Code_geo_sort'] = pd.to_numeric(df['Code_geographique'], errors='coerce')
    df = df.sort_values(by='Code_geo_sort').drop(columns='Code_geo_sort')
    return df
population_regions = sort_by_code_geo(population_regions)
menages_regions = sort_by_code_geo(menages_regions)
```
